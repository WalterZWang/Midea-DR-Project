## This demo code is to get the weather forecast data at a certain timestamp point with following parameters:
+ Database: moserveribms
+ Measurement: weather_forecast
+ Field: DSWRF_fore, RH_fore, T_fore
+ Tag: None
+ Timestemp: 2022-12-02 10:00:00
+ Whether it is a forecast data type: True
+ forecast horizon: 47 hours

```python
import pandas as pd
from Influxdb_API import ClientInfluxdb
```

```python
influxDatabase = 'moserveribms'# database name
measurement_name='weather_forecast'  #denote the measurement name
field_list=['DSWRF_fore','RH_fore','T_fore'] #denote which fields you are going to acquire
tag_dict=None
timestamp=pd.to_datetime('2022-12-02 10:00:00') #denote a certain timestamp point
fore_horizon=47 #denote forecast horizon
```

```python
db_client = ClientInfluxdb(db_name=influxDatabase) #connect the database
```

```python
df=db_client.read_influxdb(measurement_name=measurement_name,
                          start_time=timestamp, # since we can only forecast data at a certain timestamp point, 
                          end_time=timestamp,   # so start_time should be equal to end_time
                          field_list=field_list,
                          tag_dict=tag_dict,
                          fore=True,
                          fore_horizon=fore_horizon)
```
