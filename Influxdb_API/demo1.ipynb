## This demo is to get the data with following parameters:
+ Database: moserver
+ Measurement: modata
+ Field: roomTemp
+ Tag: nid = 'vrf/vrf_0000CC311178CCM26232341000271K0V/indoor/0'
+ Timestemp range: between '2022-11-29 00:00:00' and '2022-11-30 00:00:00'
+ Whether it is a forecast data type: False

```python
import pandas as pd
from Influxdb_API import ClientInfluxdb
```

```python
influxDatabase = 'moserver'
measurement_name = 'modata'
field_list = ['roomTemp']
tag_dict = {'nid': 'vrf/vrf_0000CC311178CCM26232341000271K0V/indoor/0'}
start_t = pd.to_datetime('2022-11-29 00:00:00')
end_t = pd.to_datetime('2022-11-30 00:00:00')
```

```python
db_client = ClientInfluxdb(db_name=influxDatabase) # connect the database
```

```python
df = db_client.read_influxdb(measurement_name=measurement_name,
                               start_time=start_t,
                               end_time=end_t,
                               field_list=field_list,
                               tag_dict=tag_dict,
                               fore=False)
```
