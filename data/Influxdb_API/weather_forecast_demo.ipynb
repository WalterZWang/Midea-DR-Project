## This demo code is to get the weather forecast data at a certain timestamp point with following parameters:
+ Database: moserveribms
+ Measurement: weather_forecast
+ Field: DSWRF_fore, RH_fore, T_fore
+ Tag: None
+ Timestemp: 2022-12-02 10:00:00
+ Whether it is a forecast data type: True
+ forecast horizon: 47 hours

```python
import pandas as pd
from Influxdb_API import ClientInfluxdb
```

```python
db_name='moserveribms' # database name

params={
    'measurement_name':'weather_forecast', #denote the measurement name

    'start_time':pd.to_datetime('2022-12-02 10:00:00'),# since we can only read forecast data at a certain timestamp point
    'end_time':pd.to_datetime('2022-12-02 10:00:00'), # so start_time should be equal to end_time

    'field_list':['DSWRF_fore','RH_fore','T_fore'], #denote which fields you are going to retrieve
    'tag_dict':None,
    'fore':True,
    'fore_horizon':47, # #denote forecast horizon
    'interval':'60m' #'60m' means the forecast interval is 60 minutes

}
```

**interval**: this parameter is not None only when we want to read the forecast data type, that is **fore**=True, otherwise it will be default None. Besides, the **interval** parameter shoud be input as the  "_ _m"  sting format, such as '5m', '15m', '30m', '60m', '120m', otherwise it will lead to errors.

```python
db_client = ClientInfluxdb(db_name=db_name) #connect the database
```

```python
df=db_client.read_influxdb(**params)
```
