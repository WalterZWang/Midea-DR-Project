+ Measurement: modata
+ Field: roomTemp

+ Timestemp range: between '2022-11-29 00:00:00' and '2022-11-30 00:00:00'

```python
import pandas as pd
from Influxdb_API import ClientInfluxdb
```

```python
db_name='moserver'
params={
    'measurement_name':'modata',
    'start_time':pd.to_datetime('2022-11-29 00:00:00'),
    'end_time':pd.to_datetime('2022-11-30 00:00:00'),
    'field_list':['roomTemp'],
    'tag_dict':{'nid': 'vrf/vrf_0000CC311178CCM26232341000271K0V/indoor/0'},
    'fore':False,
    'fore_horizon':None,
    'interval':None
}
```

```python
df = db_client.read_influxdb(**params)
```
