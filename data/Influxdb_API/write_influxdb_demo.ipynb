## This demo code is to write the open_loop control forecast data at a certain timestamp point to the database with following parameters:
+ **db_name**:  moserver   $\ $ #database name
+ **measurement_name**: open_loop_control   $\ $  #measurement
+ **data**: the dataframe with the field name 'indoor_state' with 23 hours prediction horizon
+ **tag_list_in_data**: None
+ **add_in_tags_dict**: {'nid':'vrf/vrf_0000CC311178CCM26232341000271K0V/indoor/0'}
+ **fore**: True $\ $ #Whether it is a forecast data type
+ **interval**: '60m' $\ $ #forecast interval, '60m' means 60minutes.

**tag_list_in_data** indicates the tag columns in the dataframe, if the input dataframe doesn't have tag columns, it should be None

**add_in_tag_dict** indicates the tag key and the corresponing tag value you want to add in

**interval**: this parameter is not None only when we want to write the forecast data type, that is **fore**=True, otherwise it will be default None; and it should be noted that we will automatically add a tag with key='interval' and value='60m'(in this case) , that is {'interval':'60m'} in the measurement. What's more, the **interval** parameter shoud be input as the  "_ _m"  sting format, such as '5m', '15m', '30m', '60m', '120m', otherwise it will lead to errors，be carefully!

```python
import pandas as pd
import numpy as np
from Influxdb_API import ClientInfluxdb
```

```python
### import the demo data to be written in the database
data_df=pd.read_csv('./indoor_state_forecast.csv',index_col=0)
data_df
```

```python
db_name='moserver'

params={'measurement_name':'open_loop_control',
        'data':data_df,
        'tag_list_in_data':None,
        'add_in_tag_dict':{'nid':'vrf/vrf_0000CC311178CCM26232341000271K0V/indoor/0'},
        'fore': True,
        'interval':'60m'
}
```

```python
db_client = ClientInfluxdb(db_name=db_name) #connect the database
```

```python
db_client.write_influxdb(**params)
```
