# Visualization of indoor temperature and control commands

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from Influxdb_API.Influxdb_API import ClientInfluxdb
```

```python
Tset = 22


for i in range(7):

    db_name='moserver'
    params={
        'measurement_name':'RBC',
        'start_time':pd.to_datetime('2023-4-8 00:00'),
        'end_time':pd.to_datetime('2023-4-9 00:00'),
        'field_list':['roomTemp', 'onOff', "runMode", 'tempSetting'],
        'tag_dict':{'nid': 'vrf/vrf_0000CC311178CCM26232341000271K0V/indoor/{}'.format(i)},
        'fore':False,
        'fore_horizon':None,
        'interval':None
    }

    db_client = ClientInfluxdb(db_name=db_name)  # connect the database
    df = db_client.read_influxdb(**params)
    df = df.dropna(subset=['runMode'], how='all')
    df = df.fillna(method='ffill')
    sns.set(rc={'figure.figsize': (10, 4)})
    sns.set_style("white", {'xtick.bottom': True})

    # 创建第一个Axes对象，并在其上绘制第一条折线
    ax1 = sns.lineplot(data=df['onOff'], label='onOff', color=sns.color_palette('Set2')[1], linewidth=2, legend=False)
    ax1.set_xticks(df.index)
    ax1.set_xticklabels(df.index.strftime('%H:%M'), rotation=90)
    # 获取第一条折线的图例信息
    handles, labels = ax1.get_legend_handles_labels()
    # 创建第二个Axes对象，并在其上绘制第二条折线
    ax2 = ax1.twinx()
    sns.lineplot(data=df['roomTemp'], color=sns.color_palette('Set2')[2], ax=ax2, label='roomTemp', linewidth=2)
    ax2.axhline(y=Tset-1, color='gray', linestyle='--')
    ax2.axhline(y=Tset+1, color='gray', linestyle='--')
    ax2.axhspan(Tset-1, Tset+1, alpha=0.5, color=sns.color_palette('Set2')[0])
    # 将第一条折线的图例信息传递给第二个Axes对象
    handles2, labels2 = ax2.get_legend_handles_labels()
    handles += handles2
    labels += labels2
    ax2.legend(handles, labels, loc='best')
    plt.title('Indoor Unit {}'.format(i))
    plt.tight_layout()
    # plt.savefig('Indoor Unit {}.png'.format(i), dpi=200)
    plt.show()
    print(df)
```
