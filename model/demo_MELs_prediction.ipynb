## The MELsModel class with heuristic method has two ways to use

### 1. Just use it for prediction, which is also the way we will use **<font color = red>most frequently</font>** later in open loop control
+ On this conditon, we should set parammeter **validate**=False

+ Set **method**='heuristic'

+ The paramter **timeStamp** means you want the future 24 hours prediction after this **timeStamp**. For example, if **timeStamp**=pd.to_datetime(**<font color = red>'2022-11-25 01:00:00'</font>**), the model will return the prediction value of **<font color = red>'2022-11-25 02:00:00','2022-11-25 03:00:00', ...... ,'2022-11-26 01:00:00'</font>** , which are total 24 time steps. We will return a dataframe with time index

```python
MELs_model=MELsModel()

y_pred_df=MELs_model.one_day_MELs_predict(method='heuristic',
                                          timeStamp=pd.to_datetime('2022-11-25 01:00:00'),
                                          validate=False)

y_pred_df
```

### 2. When we set parameter **validate**=True, we can validate the heuristic model, this application only be used when we are developing the  heuristic model, we want to know its performance. It will return dataframe and square_error of this timestamp prediction

```python
MELs_model=MELsModel()

y_pred_df,square_error=MELs_model.one_day_MELs_predict(method='heuristic',
                                          timeStamp=pd.to_datetime('2022-11-25 01:00:00'),
                                          validate=True)

y_pred_df
```

```python
square_error**0.5
```


#### Example code： get heuristic model rmse
This is not a formal way to evaluate a model, but we can get a rough idea about the permance of heuristic model

```python
# evaluation data time range
t_range=pd.date_range(pd.to_datetime('2022-08-01 00:00:00'),pd.to_datetime('2022-11-25 00:00:00'),freq='1D')
square_errors=[]

MELs_model=MELsModel()



for t in tqdm(t_range):

    _,square_error=MELs_model.one_day_MELs_predict(method='heuristic',
                                      timeStamp=t,
                                      validate=True)
    square_errors.append(square_error)








```

```python
rmse=np.nanmean(square_errors)**0.5
rmse
```
